```python
# Q4 — Count Frequency of Numbers
# numbers = [1,2,2,3,3,3,4,4,4,4]
# Create frequency dictionary.

numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]   

frequency = {}
for num in numbers:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1

print("Numbers  :", numbers)
print("Frequency:", frequency)
```

```python
# Q5 — Word Counter
# Take sentence input and count word frequency.

sentence = input("Enter a sentence: ")
words = sentence.split()

word_count = {}
for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

print("Word frequency:", word_count)
```

```python
# Q6 — Student Marks & Average
# marks = {"Math":80, "English":70, "Physics":90}
# Calculate total and average.

marks = {"Math": 80, 
        "English": 70, 
        "Physics": 90
        }

total = 0
for subject, score in marks.items():
    total += score

average = total / len(marks)

print("Marks  :", marks)
print("Total  :", total)
print("Average:", average)
```

```python
# Q7 — Highest Scoring Student
# students = {
# "Ali":85,
# "Sara":92,
# "Ahmed":78,
# "Zara":95
# }
# Find highest scoring student.

students = {
    "Ali": 85,
    "Sara": 92,
    "Ahmed": 78,
    "Zara": 95
}

highest = 0
name = ""

for student, marks in students.items():
    if marks > highest:
        highest = marks
        name = student

print("Highest scoring student:", name)
print("Marks:", highest)
```

```python
# Q8 — Merge Dictionaries
# dict1 = {"a":1,"b":2}
# dict2 = {"c":3,"d":4}
# Merge and print result.

dict1 = {"a": 1, "b": 2}
dict2 = {"c": 3, "d": 4}

merged = {}
for key, value in dict1.items():
    merged[key] = value
for key, value in dict2.items():
    merged[key] = value

print("dict1 :", dict1)
print("dict2 :", dict2)
print("Merged:", merged)
```
