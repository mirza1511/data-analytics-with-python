```python
# 1. Write a program that accepts a list from user and print the alternate element of list.

n = int(input("How many elements are in the list? "))
my_list = []

for i in range(n):
    value = input("Enter element: ")
    my_list.append(value)

print("Alternate elements:")
for i in range(0, len(my_list), 2):
    print(my_list[i])
```

```python
# 2. Write a program that accepts a list from user. Your program should reverse the content of list and display it. Do not use reverse() method.

n = int(input("How many elements are in the list? "))
my_list = []

for i in range(n):
    my_list.append(int(input("Enter element: ")))

reversed_list = []
for i in range(len(my_list) - 1, -1, -1):
    reversed_list.append(my_list[i])

print("Original list:", my_list)
print("Reversed list:", reversed_list)
```

```python

# 3. Find and display the largest number of a list without using built-in function max(). Your program should ask the user to input values in list from keyboard.

n = int(input("How many elements are in the list? "))
my_list = []

for i in range(n):
    my_list.append(int(input("Enter number: ")))

largest = my_list[0]
for num in my_list:
    if num > largest:
        largest = num

print("Largest number:", largest)
```

```python
# 4. Write a program that rotates the element of a list so that the element at the first index moves to the second index, the element in the second index moves to the third index, etc., and the element in the last index moves to the first index.

n = int(input("How many elements are in the list? "))
my_list = []

for i in range(n):
    my_list.append(int(input("Enter element: ")))

print("Before rotation:", my_list)

last = my_list[-1]
for i in range(len(my_list) - 1, 0, -1):
    my_list[i] = my_list[i - 1]
my_list[0] = last

print("After rotation:", my_list)
```

```python
# 5. Write a program that input a string and ask user to delete a given word from a string.

text = input("Enter a sentence: ")
word = input("Which word do you want to delete? ")

words = text.split()
new_words = []

for w in words:
    if w != word:
        new_words.append(w)

result = " ".join(new_words)
print("Result:", result)
```

```python
# 6. Write a program that reads a string from the user containing a date in the form mm/dd/yyyy. It should print the date in the form March 12, 2021.

months = ["January", "February", "March", "April", "May", "June",
          "July", "August", "September", "October", "November", "December"]

date = input("Enter date (mm/dd/yyyy): ")
parts = date.split("/")

month = int(parts[0])
day = int(parts[1])
year = parts[2]

print(f"{months[month - 1]} {day}, {year}")

```

```python
# 7. Write a program with a function that accepts a string from keyboard and create a new string after converting character of each word capitalized. For instance, if the sentence is "stop and smell the roses." the output should be "Stop And Smell The Roses"

def capitalize_words(sentence):
    words = sentence.split()
    new_words = []
    for w in words:
        new_words.append(w.capitalize())
    return " ".join(new_words)


text = input("Enter a sentence: ")
text = text.rstrip(".")          
print(capitalize_words(text))
```

```python
# # 8. Find the sum of each row of matrix of size m x n. For example for the following matrix output will be like this :

# #     2   11   7   12
# #     5   2    9   15
# #     8   3    10  42

# #    Sum of row 1 = 32
# #    Sum of row 2 = 31
# #    Sum of row 3 = 63


m = int(input("Number of rows: "))
n = int(input("Number of columns: "))

matrix = []
for i in range(m):
    row = []
    for j in range(n):
        row.append(int(input(f"Element [{i}][{j}]: ")))
    matrix.append(row)

for i in range(m):
    total = 0
    for j in range(n):
        total = total + matrix[i][j]
    print("Sum of row", i + 1, "=", total)
```

```python
# 9. Write a program to add two matrices of size n x m.

def read_matrix(rows, cols, name):
    matrix = []
    for i in range(rows):
        row = []
        for j in range(cols):
            row.append(int(input(f"{name}[{i}][{j}]: ")))
        matrix.append(row)
    return matrix


n = int(input("Number of rows: "))
m = int(input("Number of columns: "))

print("Enter Matrix A:")
A = read_matrix(n, m, "A")
print("Enter Matrix B:")
B = read_matrix(n, m, "B")

result = []
for i in range(n):
    row = []
    for j in range(m):
        row.append(A[i][j] + B[i][j])
    result.append(row)

print("A + B =")
for row in result:
    print(row)
```

```python
# 10. Write a program to multiply two matrices

def read_matrix(rows, cols, name):
    matrix = []
    for i in range(rows):
        row = []
        for j in range(cols):
            row.append(int(input(f"{name}[{i}][{j}]: ")))
        matrix.append(row)
    return matrix


r1 = int(input("Rows of Matrix A: "))
c1 = int(input("Columns of Matrix A: "))
r2 = int(input("Rows of Matrix B: "))
c2 = int(input("Columns of Matrix B: "))

if c1 != r2:
    print("Multiplication not possible! Columns of A must equal rows of B.")
else:
    print("Enter Matrix A:")
    A = read_matrix(r1, c1, "A")
    print("Enter Matrix B:")
    B = read_matrix(r2, c2, "B")

    result = []
    for i in range(r1):
        row = []
        for j in range(c2):
            total = 0
            for k in range(c1):
                total = total + A[i][k] * B[k][j]
            row.append(total)
        result.append(row)

    print("A x B =")
    for row in result:
        print(row)
```
