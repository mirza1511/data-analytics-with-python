6) Add a new column 'Bonus' that is 10% of Salary and a column 'Total' = Salary + Bonus. Then rename the column 'City' to 'Location' and drop the 'Bonus' column.

```python
import pandas as pd

data = {
    "Name": ["Ali", "Ahmed", "Sara", "Ayesha", "Bilal", "Usman"],
    "Age": [22, 35, 31, 24, 42, 23],
    "City": ["Karachi", "Delhi", "Delhi", "Karachi", "Delhi", "Multan"],
    "Salary": [30000, 45000, 35000, 45000, 55000, 32000]
}

df = pd.DataFrame(data)
print(df)

df["Bonus"] = df["Salary"] * 0.10

df["Total"] = df["Salary"] + df["Bonus"]

df = df.rename(columns={"City": "Location"})

df = df.drop(columns=["Bonus"])

print()
print(df)
```

7) Create a DataFrame having some missing values (NaN). Find the count of missing values in each column, fill numeric NaNs with the column mean, and drop rows that still have missing values.

```python
import pandas as pd

data = {
    "Name": ["Ali", "Ahmed", None, "Ayesha", "Bilal"],
    "Age": [20, None, 25, 22, None],
    "Salary": [30000, 40000, None, 50000, 45000]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

print("\nMissing Values:")
print(df.isnull().sum())

df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

print("\nAfter Filling Numeric Missing Values:")
print(df)

df = df.dropna()

print("\nAfter Dropping Remaining Missing Rows:")
print(df)
```

8) Sort the DataFrame by Salary in descending order, and then by Age in ascending order. Also print the top 3 highest-paid employees using nlargest().

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Ali', 'Sara', 'Ahmed', 'Ayesha', 'Usman'],
    'Age': [25, 30, 28, 24, 30],
    'Salary': [50000, 70000, 60000, 70000, 80000]
})

print("Original Data:")
print(df)

sorted_df = df.sort_values(
    by=['Salary', 'Age'],
    ascending=[False, True]
)

print("\nSorted DataFrame:")
print(sorted_df)

print("\nTop 3 Highest-Paid Employees:")
print(df.nlargest(3, 'Salary'))
```

9) Using groupby, find the average salary per City, the number of employees per City, and the maximum salary per City. Show the result using agg() with multiple functions.

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Ali', 'Sara', 'Ahmed', 'Ayesha', 'Usman'],
    'City': ['Karachi', 'Lahore', 'Karachi', 'Lahore', 'Islamabad'],
    'Salary': [50000, 70000, 60000, 80000, 90000]
})

print("Original Data:")
print(df)
print()

result = df.groupby('City').agg(
    Average_Salary=('Salary', 'mean'),
    Employee_Count=('Name', 'count'),
    Maximum_Salary=('Salary', 'max')
)

print(result)
```

10) Create two DataFrames (employees and departments) with a common column 'DeptID'. Merge them using inner and left joins, then concatenate two DataFrames of the same structure using pd.concat() and remove duplicate rows.

```python
import pandas as pd

employees = pd.DataFrame({
    'Name': ['Ali', 'Sara', 'Ahmed', 'Ayesha'],
    'DeptID': [1, 2, 1, 3]
})

departments = pd.DataFrame({
    'DeptID': [1, 2, 4],
    'Department': ['IT', 'HR', 'Finance']
})

inner_join = pd.merge(employees, departments, on='DeptID', how='inner')
print("Inner Join:")
print(inner_join)

left_join = pd.merge(employees, departments, on='DeptID', how='left')
print("\nLeft Join:")
print(left_join)

df1 = pd.DataFrame({
    'Name': ['Ali', 'Sara'],
    'DeptID': [1, 2]
})

df2 = pd.DataFrame({
    'Name': ['Sara', 'Ahmed'],
    'DeptID': [2, 1]
})

combined = pd.concat([df1, df2], ignore_index=True)

combined = combined.drop_duplicates()

print("\nConcatenated DataFrame without duplicates:")
print(combined)
```
